## __RNA-seq Download & S3 Upload__

Streams each tumor/control RNA-seq file from the GDC API directly into S3 (`raw/RNA_seq/Tumeric_data/` and `raw/RNA_seq/Control_data/`), then uploads the case/file metadata JSON pulled in `1-API_call.ipynb`.

### __Imports & S3 client setup__

```python
import boto3, os, sys
from dotenv import load_dotenv
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))

from src.s3_utils import RNA_seq_up_and_download

load_dotenv("project.env")
```

```python
s3 = boto3.client(
    "s3",
    aws_access_key_id     = os.getenv("AWS_ACCESS_KEY_ID"),
    aws_secret_access_key = os.getenv("AWS_SECRET_ACCESS_KEY"),
    region_name           = os.getenv("AWS_REGION")
)
```

Upload parsed RNA-seq files to S3

```python
with open(os.path.join(os.getenv("Output_path"), "Tumor_UUID_download.txt"), "r") as Tumor_files:
    Tumor_UUID = Tumor_files.readlines()

with open(os.path.join(os.getenv("Output_path"), "Control_UUID_download.txt"), "r") as Control_files:
    Control_UUID = Control_files.readlines()
    
for file in Tumor_UUID:
    RNA_seq_up_and_download("Tumeric_data", file.strip(), s3)
    
for file in Control_UUID:
    RNA_seq_up_and_download("Control_data", file.strip(), s3)
```

Upload raw files to S3

```python
s3.upload_file("data_cases.json", os.getenv("AWS_BUCKET_NAME"), "raw/clinical/data_cases.json")
s3.upload_file("data_files.json", os.getenv("AWS_BUCKET_NAME"), "raw/files/data_files.json")
```
